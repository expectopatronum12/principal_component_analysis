```python
import numpy as np
import pandas as pd
import plotly.express as px 
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import os
print(f"all modules have been imported!")
```

```python
# lets take the data from the actual japanese website that we have downloaded as an excel.
target_filename = "japan_cpi_data.xlsx"
full_path = os.getcwd() + "\\" + target_filename
target_df = pd.read_excel(full_path, sheet_name = None)
all_keys = [key for key in target_df.keys()]
all_keys

target_df = target_df[all_keys[0]]
#target_df = target_df.iloc[10:, 7:]
#target_df.iloc[10:,7:]

target_df.columns = target_df.iloc[10,:]
target_df.iloc[13:692, 7:]


#target_df.columns
```

```python
np.random.seed(42)
mock_subcomponents = ['food', 'energy', 'housing', 'transport', 'medical', 'clothing']
df_subcomponents = pd.DataFrame(
    np.random.rand(100,6),
    columns = mock_subcomponents
)

df_subcomponents

# Clean out missing data
df_clean = df_subcomponents.dropna()
df_clean
```

```python
# 2. Standardize the data
# PCA is scale-sensitive. if Energy moves 10% but housing moves 1%, 
# PCA will think Energy is more important just because the numbers are bigger

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_clean)
X_scaled
```

```python
# 3. Compute the eigenvectors and eigenvalues

pca = PCA()
pca.fit(X_scaled)
```

```python
# Eigenvalues represent the variance explained by each axis

eigenvalues = pca.explained_variance_
variance_ratio = pca.explained_variance_ratio_

# Eigenvectors represent the mathematical direction/weights of the subcomponents.
eigenvectors = pca.components_
```
